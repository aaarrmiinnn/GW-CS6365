1. Generalized Linear Model (GLM).
    * Purpose: Captures lower-order interactions, such as simple linear relationships.
    * Sample Input: Features = $[Age: 30, Gender: Male, Income: $50k]
    * Sparse Features (One-hot encoded categories) = [0, 1, 0, 0, 1, 0]
    * Process: Weighted linear combination of input features: $𝑂𝑢𝑡𝑝𝑢𝑡 = 𝑤1 ⋅ 𝐴𝑔𝑒 + 𝑤2 ⋅ 𝐺𝑒𝑛𝑑𝑒𝑟 + 𝑤 * 𝐼𝑛𝑐𝑜𝑚𝑒$ 
    * Output=$w 1 ​ ⋅Age+w 2 ​ ⋅Gender+w 3 ​Income$ = 0.7 (e.g., likelihood of clicking on an item)
2. Wide & Deep Models (GLM + MLP)
    * Purpose: Combines lower-order (GLM-style) and higher-order (deep neural net) feature interactions.
    * Sample Input:
    * Features: $[Age: 30, Gender: Male, Income: $50k]  
    * Sparse + Dense Encoded Embeddings: [0, 1, 0, 0, 1, 0, 0.2, 0.8]
    * Process Wide (GLM): Applies linear weights to the input features (same as GLM).  
    * Process Deep (MLP): Embeddings are passed through hidden layers to learn complex patterns like $( \text{Age} \times \text{Income} )$.
3. Multi-Layer Perceptron (MLP)
    * Purpose:  Captures higher-order interactions without relying on linearity assumptions.
    * Sample Input: Dense Encoded Features = [0.2, 0.5, 0.7, 0.9] (e.g., learned embeddings of user and item features).
    * Process: 
      - Entire input passes through multiple dense layers, applying activation functions (e.g., ReLU).  
      - Captures complex relationships like $( \text{Age}^2 \times \text{Income} )$.
    * Sample Output:
        $( \text{Output} = 0.9 )$ (e.g., likelihood of watching a recommended video).

DeepFM is a model that combines the strengths of **Factorization Machines (FM)** and **Deep Neural Networks (DNNs)** to capture both low-order and high-order feature interactions in recommendation systems.

---

## Key Components of DeepFM

1. **Sparse Features**:
   - Represent the input features (e.g., user ID, item ID, categorical data).
   - These are mapped into dense embeddings.
   
2. **Dense Embeddings**:
   - Transform sparse inputs into dense representations for easier modeling of interactions.

3. **FM Layer**:
   - Captures low-order feature interactions using the inner product between embeddings.

4. **Deep Neural Network (DNN)**:
   - Captures high-order feature interactions by passing embeddings through multiple dense layers.

5. **Output Unit**:
   - Combines the outputs of the FM and DNN parts to make the final prediction.

---

## Example

### Sample Input:
- **User Features**: Age = 25, Gender = Male
- **Item Features**: Category = Electronics, Brand = Sony

---

### Step-by-Step Process

1. **Sparse to Dense Embeddings**:
   - Age → Dense embedding: `[0.2, 0.5]`
   - Gender → Dense embedding: `[0.1, 0.8]`
   - Category → Dense embedding: `[0.3, 0.6]`
   - Brand → Dense embedding: `[0.7, 0.4]`

2. **FM Layer (Low-Order Interactions)**:
   - Compute pairwise inner products of the embeddings to model combinations like:
     - Age × Gender
     - Category × Brand

3. **DNN (High-Order Interactions)**:
   - Pass embeddings through hidden layers to capture complex relationships like:
     - \( (\text{Age} + \text{Gender}) \times (\text{Category} + \text{Brand}) \)

4. **Combine Outputs**:
   - Add the FM and DNN outputs.
   - Apply sigmoid activation to predict a value (e.g., likelihood of a user interacting with the item).

---

### Sample Output:
- **Output**: \( 0.85 \) (e.g., an 85% chance the user will buy a Sony electronic device).

---

## Summary
DeepFM automates feature engineering, reducing manual effort while capturing diverse feature interactions, making it highly effective for recommendation systems.

| **Aspect**              | **FM Layer**                                       | **Hidden Layers (DNN)**                         |
|-------------------------|---------------------------------------------------|------------------------------------------------|
| **Focus**               | Pairwise (low-order) interactions                 | Higher-order, non-linear interactions          |
| **Mechanism**           | Inner product of embeddings                       | Non-linear transformations using dense layers  |
| **Parameters**          | Fewer parameters, simpler structure               | More parameters, deep structure                |
| **Efficiency**          | Computationally efficient for sparse features     | More computation-intensive                     |
| **Interaction Modeling**| Explicit pairwise modeling                        | Implicit, hierarchical modeling                |
| **Use Case**            | Useful for linear, simple feature interactions    | Effective for complex, high-order patterns     |

### Key Components
1. **Embedding and Stacking Layer**:
   - Combines sparse and dense features into embeddings.
   - Initial input features are represented as $( x_0 $).

2. **Cross Network**:
   - Explicitly models feature interactions with a cross-layer formula:
     $[
     x_{l+1} = x_0 x_l^T w_l + b_l + x_l
     $]
   - Efficiently captures low-order interactions.

3. **Deep Network**:
   - Processes the embeddings through multiple dense layers (e.g., ReLU activation).
   - Learns high-order, non-linear feature interactions.

4. **Combination Layer**:
   - Merges outputs from the Cross and Deep networks.
   - Passes the combined representation to the final output layer.

5. **Output Layer**:
   - Predicts probabilities, e.g., $( p = \text{sigmoid}(W_{\text{logit}} x_{\text{stack}} + b_{\text{logit}}) )$.

---

### Advantages
- **Low Variance**:
  - Cross layers explicitly capture feature interactions, reducing the need for extensive deep layers.
- **Efficiency**:
  - Fewer parameters compared to fully connected layers, making the model computationally efficient.
- **Hybrid Capability**:
  - Combines the strengths of feature engineering (Cross network) and deep learning (Deep network).

---

### Workflow Example
1. **Input**:
   - Sparse: $[User: A, Category: Electronics]$
   - Dense: $[Age: 30, Income: $$50k]$
2. **Cross Network**:
   - Models interactions like $(\text{User: A} \times \text{Electronics}, \text{Age} \times \text{Income})$.
3. **Deep Network**:
   - Processes embeddings for complex patterns.
4. **Output**:
   - Probability, e.g., $( 0.85 \ for a user clicking on an ad.


```python
import tensorflow as tf
from tensorflow.keras.layers import Dense, Embedding, Input, Concatenate
from tensorflow.keras.models import Model

# Example inputs
dense_features = Input(shape=(3,), name="dense_features")  # e.g., Age, Price, etc.
sparse_features = Input(shape=(2,), name="sparse_features", dtype="int32")  # e.g., Categorical IDs

# Embedding for sparse features
embedding_layer = Embedding(input_dim=100, output_dim=8)  # Assume 100 categories
embedded_sparse = embedding_layer(sparse_features)

# Flatten embedding
flattened_sparse = tf.reshape(embedded_sparse, [-1, 16])  # Combine embedding dimensions

# Combine dense and sparse features
x0 = Concatenate()([dense_features, flattened_sparse])

# Deep network
deep = Dense(128, activation="relu")(x0)
deep = Dense(64, activation="relu")(deep)

# Cross network
cross = x0
for i in range(3):  # 3 cross layers
    cross = tf.keras.layers.Add()([
        tf.matmul(tf.expand_dims(cross, -1), tf.expand_dims(x0, -1), transpose_b=True) @ tf.Variable(tf.random.normal((x0.shape[-1], 1))),
        cross
    ])

# Combine deep and cross outputs
combined = Concatenate()([deep, cross])
output = Dense(1, activation="sigmoid")(combined)

# Build model
model = Model(inputs=[dense_features, sparse_features], outputs=output)
model.compile(optimizer="adam", loss="binary_crossentropy")
```

# What is the core issue? Practical approach

- **Challenges:**
  - Not possible to take the most accurate model and score all items due to latency issues.
  - Cannot perform large-scale exploration via brute force.

- **Solution: Heuristic-based Multi-Stage Computations:**
  1. **Stage 1:** Use simple heuristics or models to retrieve top-\(k\) candidates from billions.
  2. **Stage 2:** Re-rank the top-\(k\) using advanced modeling approaches.

- **Key Considerations:**
  - As Stage 2 models improve, Stage 1 can become more comprehensive.
  - Exploration can happen at both stages.
  - Setting up Stage 1 is more iterative and heuristic-driven than scientific.

  - **LinkedIn:** Stage 1 was comprehensive; Stage 2 improved significantly.
  - **Pinterest:** Gains largely came from Stage 1 improvements.
  - **Comcast:** Stage 1 was comprehensive; Stage 2 improved significantly.

- **General Advice:**
  - Use offline simulations to identify where gains are possible (though costly and complex).
  - Iterate on both stages.
  - **Practical Tip:** Be conservative with Stage 1 for precision and recall; make Stage 2 strong first, then expand Stage 1.


# Ranking: Multi-Objective Optimization


$\max_x \text{UserUtility}(x) \quad \text{s.t.} \quad \text{AdvUtility}(x) \geq \theta
$

$
\Longrightarrow \quad \max_x \text{UserUtility}(x) + w \cdot \text{AdvUtility}(x)
$

- Use a simple weighted sum of utility terms to achieve Pareto optimality (via Lagrange duality).
- Tune weights to achieve desired trade-offs.
- Real systems introduce non-linearities, such as:
  - Ad spend diversity.
  - Budget pacing.
  - Content variety for users.


# Practical considerations for Multi-objective

- **Abused in practice (watch out for it), especially for funnel optimization**
  - E.g., Maximize connections and valuable connections are not at odds with each other (shouldn’t we optimize total valuable connections? )
  - No point in maximizing clicks but reduce user satisfaction later in the funnel when it is time to buy (shouldn’t we be simply optimizing total user satisfaction? )
  - Better to spend the time in such cases to formulate the right long-term objective based on LTV.
  - Every guardrail added has a cost - need to be thoughtful when a constraint is necessary.
  - Multi-objective is not a short-cut to solve real product problems (avoid lazy approach).

- **Multi-objective truly useful when utilities of two different entities in a marketplace are competing with each other (e.g., advertising and engagement).**

# Simple Example: Most Popular with 2 Items

### Problem
- Maximize clicks when recommending the most popular of two items.

### Oracle Solution
- If we knew the Click-Through Rates (CTR), serve the best one all the time.

### Simplified Problem
- CTR of one item (A) is known (\(q\), say 5%) from historical data.
- CTR of the other item (B) is unknown.

### Start with Random Serves
- Start with 100 random serves of item B: **3 clicks out of 100 serves.**

### Question:
- Should we go **greedy** (always serve item A)?
- Is that optimal? Why or why not?

# Intuition: Look at Future Potential for Long-Term

### Two Scenarios:
1. **Single Decision**: Give 100% to Item A.
2. **Multiple Decisions**:
   - Explore Item B because its CTR can potentially be higher.

### Formula for Potential
$
\text{Potential} = \int_{p > q} (p - q) \cdot f(p) \, dp
$

Where:
- \( q \): CTR of Item A.
- \( p \): CTR of Item B.
- \( f(p) \): Probability density function of Item B's CTR.

# Back to our 2-item example: Approach 1

---

### Solution principle - Optimism in the face of uncertainty

Use optimistic estimate:
$
p(t) = \frac{c(t)}{n(t)} + \sqrt{\frac{\log(t)}{n(t)}}
$

- We keep playing **A** until we are sure it is worse with high probability.
- This is also called the **Upper Confidence Bound (UCB) algorithm**.

---

### Properties

- **Typically explores more than other competitive methods.**
- **Not robust** to breakdown of some assumptions (e.g., delayed feedback).
- **Not trivial** to generalize to personalization (but can/has been done).
- **Easy to implement**: No need to estimate anything beyond point estimates/scores (for personalization, variance is needed).

- **Not an MAB method**, just a good heuristic. More robust to breakdown in assumptions and more broadly applicable.


# Multi-Armed Bandits: Two Approaches

---

### MDP approach (Needs estimates of distribution)
- Seeks to find the Bayes optimal solution to a **Markov Decision Process (MDP)** with assumptions about probability distributions.
- **Representative work**: Gittins' index, Whittle's index.
- **Computationally intensive**.

---

### Minimax approach (Needs only point estimates)
- Seeks to find a scheme that incurs **bounded regret** (with no or mild assumptions about probability distributions). At best logarithmic.
- **Representative work**: UCB by Lai, Auer.
- **Usually computationally easy**, but they tend to explore more than needed in practice.

# Do Classical MAB apply to Recommender Systems? NO

- **Item pool changes**, utilities like CTRs are not stationary, feedback is not immediate.
- **Most important**: Arms are dependent. Best arm depends on the user (so each user is a separate MAB in some sense).
- **Multiple objectives**: Revenue, clicks, user retention, freshness, etc.
- **Why keep MAB in mind?**
  - Helps us build better solutions.
  - **ML models** reduce the dimensionality of the explore/exploit problem. This is key to practical solutions.


# Softmax

- **What if no posterior distribution?** You just have scores/point estimates.
- Same idea as Thompson Sampling: **Sample arm \(i\)** with probability:
 $
  \frac{\exp\{\hat{\mu}_i / \tau\}}{\sum_j \exp\{\hat{\mu}_j / \tau\}}, \quad \hat{\mu}_i = \text{predicted CTR of item } i, \, \tau = \text{temperature}
$
- Works well in low dimensions (like 100s of arms). Needs effort to tweak $\tau$.
- **Not good in high dimensions**.
- **Robust** like Thompson Sampling with only one tweaking constant.
- Easy to generalize to personalization (just need scores).


```python
import numpy as np

# Predicted CTRs (scores) for three movies
movie_scores = np.array([0.8, 0.6, 0.4])  # Movie A, B, C

# High temperature (encourages exploration)
tau = 2.0

# Softmax function
def softmax(scores, temperature):
    scaled_scores = scores / temperature
    exp_scores = np.exp(scaled_scores)
    probabilities = exp_scores / np.sum(exp_scores)
    return probabilities

# Calculate probabilities with high temperature
probabilities = softmax(movie_scores, tau)

# Print probabilities
print("Movie Scores:", movie_scores)
print("Probabilities (High Exploration):", probabilities)

# Simulate movie selection based on probabilities
selected_movie = np.random.choice(['Movie A', 'Movie B', 'Movie C'], p=probabilities)
print("Selected Movie:", selected_movie)
```

# epsilon-greedy

- With small probability epsilon choose random; else, choose the best empirically estimated arm.
- Simple but **over-explores**:
  - **Mitigation**: Reduce epsilon monotonically but still explores more than needed.
- **Robust**: Easily generalizes to personalization.
- **Practice**:
  - Randomize for a small fraction epsilon of users.
  - This randomized traffic has benefits like offline evaluation and long-term impact study.

# Reducing Dimensionality for Explore/Exploit in RS

- Need to run explore/exploit (e/e) for each user: **Millions of users** and billions of items.
- Focus where it matters: **Cold-start scenarios**.
- Approach: Reduce dimensionality of e/e via supervised learning.
  - Do e/e in the lower dimension:
    $
    \Pr(\text{action} | \text{user } i, \text{item } j, \text{context } k) = h(x_r, z_r, v_k, \Theta)
    $
  - Explore/exploit for $(\Theta)$.
- If we have posterior of $(\Theta)$: Sample as in Thompson Sampling.
- **epsilon-greedy**: Works well if the model is good.
- **UCB**: Works when variance estimates are easily available (like linear \(h\)).
- **Softmax**: Works but unpredictable in high dimensions.


# Practical Advice Based on Experience

- Focus on building models for **cold-start**; use simple e/e methods that align with your modeling approach.
- Offline evaluation is crucial; supervised learning eval methods don’t work for RS systems.
- Randomize a small traffic of users for explore/exploit; leave the rest for exploitation.
  - **Example**: epsilon-greedy is easy to debug for small traffic.
- For better cold-start, try **Bayes-optimal explore/exploit** for segmented user clusters.
  - Can use decision trees or ensemble approaches.


```python
# Step 1: Import necessary libraries
import pandas as pd
import numpy as np

# Step 2: Define mock data for menus
data = {
    "Menu": ["Menu A", "Menu B", "Menu C", "Menu D"],
    "Clicks": [30, 10, 5, 1],  # Successes (number of clicks)
    "Impressions": [100, 50, 20, 5],  # Attempts (number of impressions)
    "Personalization_Score": [0.8, 0.6, 0.4, 0.2],  # Base scores from personalization model
}

# Step 3: Create a DataFrame
df = pd.DataFrame(data)

# Step 4: Add Beta distribution parameters for Thompson Sampling
df["a"] = df["Clicks"] + 1  # Prior successes
df["b"] = df["Impressions"] - df["Clicks"] + 1  # Prior failures

# Step 5: Simulate Thompson Sampling
np.random.seed(42)  # For reproducibility
df["TS_Sample"] = np.random.beta(df["a"], df["b"])  # Draw samples from Beta distribution

# Step 6: Combine personalization scores and TS samples
lambda_factor = 0.4  # Weight for exploration
df["Final_Score"] = df["Personalization_Score"] + lambda_factor * df["TS_Sample"]

# Step 7: Sort menus by final score
df_sorted = df.sort_values(by="Final_Score", ascending=False)

# Display the result
from IPython.display import display
display(df_sorted)

# Print the sorted DataFrame for further clarity
print(df_sorted)

```
