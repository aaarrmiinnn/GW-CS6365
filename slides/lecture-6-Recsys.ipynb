# Learning to rank

- Machine learning problem: goal is to construct a ranking model from training data
- Training data can be a partial order or binary judgments (relevant/not relevant)
- Resulting order of the items typically induced from a numerical score
- Learning to rank is a key element for personalization
- You can treat the problem as a standard supervised classification problem

# Learning to rank - Metrics

- **Quality of ranking measured using metrics as:**
  - Normalized Discounted Cumulative Gain
  - Mean Reciprocal Rank (MRR)
  - Fraction of Concordant Pairs (FCP)
  - Others...
- But, it is hard to optimize machine-learned models directly on these measures (e.g., non-differentiable)
- Recent research on models that directly optimize ranking measures



## 🔹 NDCG – Normalized Discounted Cumulative Gain

Measures how well the ranking places **highly relevant items early**, using a logarithmic discount for lower ranks.

- Rewards graded relevance (e.g., 0, 1, 2, 3)
- Widely used in search and recommender systems

**Formula:**  
DCG = $\sum_{i=1}^{n} \frac{2^{rel_i} - 1}{\log_2(i + 1)}$  
NDCG = DCG / IDCG

---

### 🧠 NDCG Example

**Ground truth relevance labels:**  
Item A: 3, Item B: 2, Item C: 0, Item D: 1, Item E: 0  
**Predicted order:** [A, B, C, D, E]  

DCG =  
$\frac{7}{\log_2(2)} + \frac{3}{\log_2(3)} + \frac{0}{\log_2(4)} + \frac{1}{\log_2(5)} + \frac{0}{\log_2(6)}$

IDCG = DCG of perfect order → [3, 2, 1, 0, 0]

NDCG = DCG / IDCG → **Score ∈ [0, 1]**

## 🔹 MRR – Mean Reciprocal Rank

Evaluates the rank position of the **first relevant item** in the result list.

- Suitable for tasks with **a single correct answer**
- Higher MRR means relevant items appear earlier

**Formula:**  
MRR = $\frac{1}{\text{rank of first relevant item}}$

---

### 🧠 MRR Example

**Predicted ranked list:**  
1. D (not relevant)  
2. B (not relevant)  
3. C (✅ relevant)  

MRR = 1 / 3 = **0.33**

For multiple queries, average the reciprocal ranks across all.



## 🔹 FCP – Fraction of Concordant Pairs

Measures the proportion of correctly ordered item pairs.

- Focuses on **pairwise correctness**
- Higher FCP → better ranking consistency

**Formula:**  
FCP = (# correct pairs) / (total possible pairs)

---

### 🧠 FCP Example

**True relevance:** A > B > C  
**Predicted order:** A > C > B

Check all pairs:  
- (A, B): correct ✅  
- (A, C): correct ✅  
- (B, C): incorrect ❌  

FCP = 2 / 3 = **0.67**


## 🔹 Precision@k

Measures the % of **top-k items** that are relevant.

- Good for evaluating shortlists or top-N recs
- Easy to interpret, but ignores order within top-k

**Formula:**  
P@k = (# relevant in top-k) / k

---

### 🧠 Precision@k Example

**Top-5 recommendation:**  
[✅, ❌, ✅, ❌, ❌] → 2 relevant out of 5

Precision@5 = 2 / 5 = **0.40**


## 🔹 Recall@k

Measures the fraction of **all relevant items** retrieved in the top-k.

- Good when there are **multiple relevant items**
- Complements Precision@k

**Formula:**  
Recall@k = (# relevant in top-k) / (total relevant)

---

### 🧠 Recall@k Example

3 relevant items exist.  
Top-5 contains 2 of them → Recall@5 = **2 / 3 = 0.67**


## 🔹 MAP – Mean Average Precision

Averages the precision **at ranks where relevant items occur**.

- Captures both **position** and **completeness**
- Used for multi-label or multi-relevant ranking

---

### 🧠 MAP Example

**Ground truth relevant items:** B, D  
**Predicted ranked list:** A, B, C, D

Now compute precision at ranks where relevant items are found:

- Rank 2: B is relevant → Precision@2 = 1/2 = 0.5  
- Rank 4: D is relevant → Precision@4 = 2/4 = 0.5

**AP = (0.5 + 0.5) / 2 = 0.5**  
(2 relevant items, so divide by 2)

**MAP = 0.5** (if only one query; else average over multiple)

# Learning to rank - Approaches

### 1. Pointwise
  - Ranking function minimizes loss function defined on individual relevance judgment
  - Ranking score based on regression or classification
  - Ordinal regression, Logistic regression, SVM, GBDT, ...

### 2. Pairwise
  - Loss function is defined on pair-wise preferences
  - Goal: minimize number of inversions in ranking
  - Ranking problem is then transformed into the binary classification problem
  - RankSVM, RankBoost, RankNet, FRank...
  - BPR is a pairwise learning to rank approach that can be applied to different methods like kNN and MF

**Learning to rank approaches:**  
https://opensourceconnections.com/blog/2017/08/03/search-as-machine-learning-prob/

| Model     | Type     | Core Method     | Loss Type       | Nonlinear | Key Strength                     |
|-----------|----------|-----------------|------------------|-----------|----------------------------------|
| RankSVM   | Pairwise | SVM-style       | Hinge            | No        | Theoretical robustness           |
| RankBoost | Pairwise | Boosting        | Exponential      | Yes       | Ensemble flexibility             |
| RankNet   | Pairwise | Neural network  | Logistic         | Yes       | Deep learning compatibility      |
| FRank     | Ordinal  | GLM             | Ordinal smooth   | Limited   | Efficiency, non-binary labels    |
| BPR       | Pairwise | MF/kNN          | Logistic         | Depends   | Implicit feedback personalization|




* You’re not just asking "is A better than B?" like in pairwise.
* You’re asking: "How do I rearrange the whole list so it's best according to some metric?

### 3. Listwise

  - **Indirect Loss Function**
    - RankCosine: similarity between ranking list and ground truth as loss function
    - ListNet: KL-divergence as loss function by defining a probability distribution
    - Problem: optimization of listwise loss function may not optimize IR metrics
  - **Directly optimizing IR metric** (difficult since they are not differentiable)
    - Genetic Programming or Simulated Annealing
    - LambdaMart weights pairwise errors in RankNet by IR metric
    - Gradient descent on smoothed version of objective function (e.g., CLiMF or TFMAP)
    - SVM-MAP relaxes MAP metric by adding to SVM constraints
    - AdaRank uses boosting to optimize NDCG
    
Direct Optimization of IR Metrics (NDCG, MAP, etc.)
    The problem: IR metrics like MAP or NDCG are non-differentiable (they’re rank-based, not smooth functions).
Solutions:

    LambdaMart: Uses RankNet's pairwise gradients, but weights them by how much a swap would change NDCG — a smart indirect way to move toward optimizing NDCG.

## Modern Graph-Based Similarity Methods

**From SimRank to Graph Neural Networks**

SimRank (2002) showed graph structure encodes similarity → GNNs (2017+) learned to extract it automatically → Simplified GNNs (2020+) proved you don't need complexity, just the right graph signal.


---

### Graph Neural Networks for Similarity

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **GraphSAGE** | 2017 | Learns node embeddings by sampling & aggregating neighbor features. Inductive: works on unseen nodes | [Hamilton et al.](https://arxiv.org/abs/1706.02216) |
| **PinSage** | 2018 | GraphSAGE at Pinterest's **3B+ node** item-item graph. Landmark GNN-in-production paper | [Ying et al.](https://arxiv.org/abs/1806.01973) |
| **LightGCN** | 2020 | Only neighborhood aggregation, no transforms. Outperforms heavier models | [He et al.(https://arxiv.org/abs/2002.02126) |


---

### Bipartite Graph Methods

For **user–item** recommendation graphs specifically.

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **NGCF** | 2019 | Propagates embeddings on user-item bipartite graph to capture collaborative signal | [Wang et al.](https://arxiv.org/abs/1905.08108) |
| **UltraGCN** | 2021 | Skips message passing — approximates infinite-layer GCN in closed form | [Mao et al.](https://arxiv.org/abs/2110.15114) |

Trend: simpler is better. LightGCN beat NGCF, UltraGCN skips propagation altogether.

---

### Knowledge Graphs for Recommendations

When you have **rich relational data** (genres, actors, directors) beyond user-item interactions.

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **KGAT** | 2019 | Attention over KG neighbors — learns which relations matter for a recommendation | [Wang et al.](https://arxiv.org/abs/1905.07854) |
| **CompGCN** | 2020 | Jointly embeds nodes and relations via composition operators inside a GCN | [Vashishth et al.](https://arxiv.org/abs/1911.03082) |

---

### Contrastive Learning on Graphs (2021–2023)

The biggest post-2021 shift. Self-supervised contrastive learning became the dominant paradigm for graph-based recs.

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **SGL** | 2021 | First to apply contrastive learning to graph recs — augments views via node/edge dropout | [Wu et al., SIGIR '21](https://arxiv.org/abs/2010.10783) |
| **SimGCL** | 2022 | Drops graph augmentation entirely — just adds noise to embeddings. Simpler and better | [Yu et al.](https://arxiv.org/abs/2112.08679) |
| **LightGCL** | 2023 | Uses SVD-based augmentation instead of random dropout. More informative contrastive views | [Cai et al.](https://arxiv.org/abs/2302.08191) |

Contrastive methods consistently beat non-contrastive GNNs (LightGCN, NGCF). They address data sparsity by creating self-supervised signals from the graph itself.

---

### LLMs Meet Graphs (2024–2025)

The newest frontier — combining large language models with graph structure.

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **LLMRec** | 2024 | Uses LLMs to augment interaction graphs with generated text features | [Wei et al., KDD '24](https://llm4graph-tutorial.github.io/) |
| **G-Refer** | 2025 | Graph retrieval-augmented LLMs — retrieves CF signals from graphs, translates to text for explainable recs | [Li et al., WWW '25](https://dl.acm.org/doi/10.1145/3696410.3714727) |

Emerging paradigms: LLMs as feature enrichers for graph nodes, GNNs as structured encoders feeding into LLMs, and graph-RAG for explainable recommendations.

---

### Timeline
```
SimRank (2002)         → Structural similarity via random walks
SimRank++ (2008)       → + edge weights, scalability
TopSim (2012)          → + top-k efficiency
GraphSAGE (2017)       → Learn embeddings via neighbor aggregation
PinSage (2018)         → GNNs at industry scale (3B+ nodes)
NGCF (2019)            → GNN on user-item bipartite graphs
KGAT (2019)            → GNN + knowledge graph attention
LightGCN (2020)        → Just aggregate, nothing else
UltraGCN (2021)        → Skip propagation entirely
SGL (2021)             → Contrastive learning on rec graphs
SimGCL (2022)          → Noise-based contrastive (no augmentation)
LightGCL (2023)        → SVD-based contrastive views
LLM + Graph (2024-25)  → LLMs as feature enrichers / explainers
```

**Takeaway:** Computing similarity → Learning it → Self-supervised learning it → Merging graph structure with LLM semantics.

## Social & Trust-Based Recommendations: Framework + Modern Evolution

### The 3-Step Framework

1. **Define trust** — What does trust mean here? (ratings, upvotes, follows, endorsements). Domain-specific and often topic-specific.
2. **Build the graph** — Users = nodes, trust scores = directed weighted edges. Trust is often asymmetric ($t_{AB} \neq t_{BA}$).
3. **Find/aggregate paths** — Compute inferred trust between non-adjacent nodes using a propagation algorithm.

---

### Classic Algorithms: How They Differ

| Algorithm | Path Strategy | Aggregation | Key Idea |
|-----------|--------------|-------------|----------|
| **TidalTrust** | Shortest paths only | Weighted average (weights = trust values) | Fast, but misses useful longer paths |
| **MoleTrust** | All paths up to depth $d$ | Simple average, removes cycles | Broader context, you choose the depth |
| **Appleseed** | All reachable (iterative) | Spreading activation with decay factor | PageRank-inspired, continuous scores for all reachable nodes |
| **Advogato** | Network flow | Max-flow capacity | Binary: "trusted or not." Community-level, not pairwise |

**What none of them handle well:** distrust propagation (enemy-of-my-enemy problem), scalability to millions of users, cold-start users with no trust edges.

---

### Key Nuances

- **Trust is often topic-specific** — I trust my colleague on programming, not cooking. Single scalar trust is a simplification.
- **The graph is incomplete and noisy** — You rarely have explicit trust. You infer it from indirect signals (upvotes, follows, interactions).
- **Chicken-and-egg** — You need interactions to compute trust, but you need trust to drive good recommendations.
- **Pipeline vs. learned** — Classic = hand-define trust → build graph → hand-designed propagation. Modern GNNs = learn the propagation function from data.

---

### Modern Evolution: Social + GNNs

| Method | Year | Key Idea | Paper |
|--------|------|----------|-------|
| **DiffNet** | 2019 | Models social influence diffusion — each GNN layer = one hop of influence propagation | [Wu et al.](https://arxiv.org/abs/1904.10322) |
| **DiffNet++** | 2020 | Unifies social influence + interest influence. Information flows between social graph and interest graph | [Wu et al.](https://arxiv.org/abs/2002.00844) |
| **SEPT** | 2021 | First contrastive learning for social recs — builds 3 views from social graph for self-supervised training | [Yu et al.](https://arxiv.org/abs/2106.03569) |
| **DGRec** | 2023 | Dynamically infers neighbor influence based on *current* interests, not static trust | [Song et al.](https://arxiv.org/abs/2304.13663) |

---

### Discussion Point for Students

Classic trust systems assumed explicit trust signals (ratings, endorsements). Modern platforms often only have implicit signals (follows, likes, watch time). How do you build trust propagation without explicit trust scores? → This is why learned embeddings replaced hand-designed trust metrics.

## Explore/Exploit

- One of the key issues when building any kind of personalization algorithm is how to trade off:
  - **Exploitation**: *Cashing in on what we know about the user right now*  
  - **Exploration**: *Using the interaction as an opportunity to learn more about the user*

- We need to have informed and optimal strategies to drive that tradeoff:
  - **Solution**: *Pick a reasonable set of candidates and show users only "enough" to gather information on them*

## Deep Learning for Content-Based Recommendations: 2014 → Now

### The Original Idea (Spotify, 2014)

**Problem:** New song uploaded, zero listens, no CF data. How to recommend it?

**Solution:** Train a CNN on mel-spectrograms to predict the 40 latent factors from Spotify's CF model. Audio → CF embedding space. Once trained, embed any new song instantly from its audio alone.

**Key insight:** Use content signals to bootstrap behavioral embeddings for cold-start items. The CF model is the teacher, the content model is the student.

---

### What's Changed

| Aspect | 2014 | Now |
|--------|------|-----|
| **Audio encoder** | Shallow CNN on mel-spectrograms | Transformer-based audio models (MERT, CLAP, MusicLM) |
| **Input modality** | Audio only | Audio + lyrics + metadata + album art + editorial tags |
| **CF target** | 40 latent factors from matrix factorization | 128-512 dim embeddings from deep two-tower or GNN models |
| **Training objective** | Regression: predict CF factors directly | Contrastive learning: co-listened songs should have similar embeddings |
| **Cold-start window** | Days to weeks | Hours — fast feedback loops shorten the gap |
| **Current challenge** | Cold-start (no data at all) | Warm-start refinement (quickly updating as initial data arrives) |

---

### Modern Audio Representation Models

| Model | Year | Key Idea | Paper |
|-------|------|----------|-------|
| **Jukebox** | 2020 | VQ-VAE that generates raw audio. Learns rich music representations as a byproduct | [Dhariwal et al., OpenAI](https://arxiv.org/abs/2005.00341) |
| **CLAP** | 2023 | Contrastive Language-Audio Pretraining. Aligns audio and text descriptions in shared embedding space | [Wu et al.](https://arxiv.org/abs/2211.06687) |
| **MERT** | 2023 | Self-supervised music understanding transformer. Pre-trained on large-scale music audio | [Li et al.](https://arxiv.org/abs/2306.00107) |
| **MusicLM** | 2023 | Google's text-to-music model. Internal representations capture genre, mood, instrumentation | [Agostinelli et al.](https://arxiv.org/abs/2301.11325) |

### From Regression to Contrastive Learning

**2014 approach (regression):**
$$\mathcal{L} = \| f_{\text{CNN}}(\text{audio}) - \vec{v}_{\text{CF}} \|^2$$
Directly minimize distance between predicted and actual CF embedding. Requires a pre-trained CF model as teacher.

**Modern approach (contrastive):**
$$\mathcal{L} = -\log \frac{\exp(\text{sim}(z_i, z_j) / \tau)}{\sum_k \exp(\text{sim}(z_i, z_k) / \tau)}$$
If two songs are co-listened, pull their content embeddings together. Push apart songs that aren't co-listened. No separate CF teacher needed — behavioral signal is baked into the contrastive objective.

**Why contrastive is better:** No dependency on a frozen CF model. Learns richer representations because it optimizes for relative similarity rather than absolute position. Naturally handles multimodal inputs (audio-text pairs via CLAP).

---

### The Insight That Hasn't Changed

Content → Behavior embedding bridging is still the core idea for cold-start. The 2014 CNN-to-CF-factors approach and the 2024 contrastive multimodal approach are solving the same problem: **if you have no user data for an item, infer where it belongs from what it is.**

What changed is the tools got better: deeper encoders, richer inputs, smarter training objectives. The architecture evolved, the insight didn't.

# Factorization Machines
Traditional matrix and tensor factorization break down when we add new data sources or continuous features, because each extension requires a brand-new algorithm. Factorization Machines solve this by providing one unified model that automatically learns interactions between any features we give it

Factorization Machines take the features you would normally pass to logistic regression and add an extra term that models pairwise interactions using latent factors. So it’s logistic regression plus factorized feature interactions.

- Generalization of regularized matrix (and tensor) factorization approaches combined with linear (or logistic) regression
- Problem: Each new adaptation of matrix or tensor factorization requires deriving new learning algorithms
    - Hard to adapt to new domains and add data sources
    - Hard to advance the learning algorithms across approaches
    - Hard to incorporate non-categorical variables


# Factorization Machines

- Approach: Treat input as a real-valued feature vector
    - Model both linear and pair-wise interaction of k features (i.e. polynomial regression)
    - Traditional machine learning will overfit
    - Factor pairwise interactions between features
    - Reduced dimensionality of interactions promote generalization
    - Different matrix factorizations become different feature representations
    - Tensors: Additional higher-order interactions
- Combines "generality of machine learning/regression with quality of factorization models"


# Factorization Machines

- Each feature gets a weight value and a factor vector
  - $O(dk)$ parameters
      - $b \in \mathbb{R}, \mathbf{w} \in \mathbb{R}^d, \mathbf{V} \in \mathbb{R}^{d \times k}$

- Model equation:

  $f(\mathbf{x}) = b + \sum_{i=1}^{d} w_i x_i + \sum_{i=1}^{d} \sum_{j=i+1}^{d} x_i x_j \mathbf{v}_i^\top \mathbf{v}_j$ - $O(d^2)$

  $= b + \sum_{i=1}^{d} w_i x_i + \frac{1}{2} \sum_{f=1}^{k} \left( \left( \sum_{i=1}^{d} x_i v_{i,f} \right)^2 - \sum_{i=1}^{d} x_i^2 v_{i,f}^2 \right)$ - $O(kd)$

```sql
Row 1: [1, 0, 0, 1, 0, 0, 0.2] (User A, Movie X, Time = 0.2)
Row 2: [0, 1, 0, 0, 1, 0, 0.6] (User B, Movie Y, Time = 0.6)
...
| User (A, B, C) | Movie (TI, NH, SW) | Time |
|----------------|--------------------|------|
| 1, 0, 0        | 1, 0, 0            | 0.2  |
| 1, 0, 0        | 0, 1, 0            | 0.6  |
| 0, 1, 0        | 0, 0, 1            | 0.61 |


# Factorization Machines (Rendle, 2010)

- **L2 regularized**
Can be used in any of
  - Regression: Optimize RMSE
  - Classification: Optimize logistic log-likelihood
  - Ranking: Optimize scores

- **Can be trained using:**
  - SGD
  - Adaptive SGD
  - ALS
  - MCMC

- **Gradient:**

  $\frac{\partial}{\partial \theta} f(\mathbf{x}) = \begin{cases}
  1 & \text{if } \theta \text{ is } b \\
  x_i & \text{if } \theta \text{ is } w_i \\
  x_i \sum_{j=1}^{d} v_{j,f} x_j - v_{i,f} x_i^2 & \text{if } \theta \text{ is } v_{i,f}
  \end{cases}$

- **Least squares SGD:**

  $\theta' = \theta - \eta \left( \left( f(\mathbf{x}) - y \right) \frac{\partial}{\partial \theta} f(\mathbf{x}) + \lambda \theta \right)$


# Factorization Machines (Rendle, 2010)

- Learning parameters:
    - Number of factors
    - Iterations
    - Initialization scale
    - Regularization (SGD, ALS) – Multiple
    - Step size (SGD, A-SGD)
    - MCMC removes the need to set those hyperparameters


### FM → Deep Learning Evolution

| Model | Year | Key Idea |
|-------|------|----------|
| **FM** | 2010 | Factorized pairwise feature interactions. The foundation |
| **Wide & Deep** | 2016 | Google: linear (wide) + DNN (deep) jointly trained |
| **DeepFM** | 2017 | FM layer + DNN. No manual feature engineering needed |
| **DCN / DCN-v2** | 2017/2021 | Cross network for explicit feature crosses + DNN. Used at Google |
| **DLRM** | 2019 | Meta's production model. Embeddings + dot-product interactions + MLP |
| **FiBiNET** | 2019 | Uses SENET to dynamically weight feature importance before interaction |

**The punchline:** Every one of these models is an answer to the
same question FM asked: "how do we model feature interactions
efficiently?" FM's flat feature vector with learned interactions
became the standard input representation for all modern
recommendation systems.

## 6. Warning: The right evaluation might matter more than the model you choose


## Offline/Online Testing Process
<img src="img/OF1.jpg" alt="Page Composition 2" style="width:100%; height:auto;" />


## Executing A/B Tests
<img src="img/OF2.jpg" alt="Page Composition 2" style="width:100%; height:auto;" />

## Offline Testing

<img src="img/OF4.jpg" alt="Page Composition 2" style="width:100%; height:auto;" />



## Offline Metrics

- For baseline metrics, prefer ranking or ranking-related metrics
- You might want to measure other aspects of the recommendation such as diversity (Maximum Marginal Relevance) or novelty
- In practice, you should keep a bank of metrics that you measure offline and over time connect to your online A/B test results, running posthoc analysis
- Find short-term surrogates that map to long-term improvements

<img src="img/OF41.jpg" alt="Page Composition 2" style="width:50%; height:auto;" />
<img src="img/OF42.jpg" alt="Page Composition 2" style="width:50%; height:auto;" />
<img src="img/OF43.jpg" alt="Page Composition 2" style="width:50%; height:auto;" />
